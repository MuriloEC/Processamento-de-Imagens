```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
imagem = cv2.imread('/content/drive/MyDrive/imagens/img_aerea.jpg')
imagem2 = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

plt.imshow(imagem)
plt.show()
```

```python
#Calcular o histograma
hist= cv2.calcHist([imagem],[0],None,[256],[0,256])

plt.plot(hist, color='green')
plt.xlabel('Intensidade de Pixel')
plt.ylabel('Frequência')
plt.title('Histograma da imagem')
plt.show()
```

```python
x_inicial = 230
y_inicial = 200
largura = 100
altura = 200

# Extraia a ROI usando NumPy slicing
roi = imagem[y_inicial:y_inicial + altura, x_inicial:x_inicial + largura]

# Exiba a ROI
plt.imshow(roi)
plt.title('Região de Interesse')
plt.show()
```

```python
hist2= cv2.calcHist([roi],[0],None,[256],[0,256])

plt.plot(hist2, color='green')
plt.xlabel('Intensidade de Pixel')
plt.ylabel('Frequência')
plt.title('Histograma da imagem')
plt.show()
```

```python
normalizado = (roi - np.min(roi)) / (np.max(roi) - np.min(roi))
```

```python
print(normalizado)
```

```python
plt.imshow(normalizado)
plt.title('Região de Interesse')
plt.show()
```

```python
# Redimensionar img2 para corresponder às dimensões de img1
img2_resized = cv2.resize(roi, (imagem.shape[1], imagem.shape[0]))

# Sobrepor as imagens com pesos
dst = cv2.addWeighted(imagem, 0.7, img2_resized, 0.3, 0)

# Exibir a imagem resultante
plt.imshow(dst)
plt.show()
```
