```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image_gray = cv2.imread('/content/drive/MyDrive/imagens/mammog.jpg', cv2.IMREAD_GRAYSCALE)

if image_gray is not None:
    plt.imshow(image_gray, cmap='gray')
    plt.title('Image Gray')
    plt.axis('off')
    plt.show()

else:
  print("Erro ao ler a imagem")
```

```python
image_invertida = 255 - image_gray
plt.imshow(image_invertida, cmap='gray')
plt.title('Imagem Invertida')
plt.axis('off')
plt.show()
```

```python
image_gray = cv2.imread('/content/drive/MyDrive/imagens/spectrum.jpg', cv2.IMREAD_GRAYSCALE)

if image_gray is not None:
    plt.imshow(image_gray, cmap='gray')
    plt.title('Image Gray')
    plt.axis('off')
    plt.show()

else:
  print("Erro ao ler a imagem")
```

```python
image_np = np.array(image_gray)
image_np = image_np+2
img_1 = np.log10(image_np)
img_2 = ((img_1-img_1.min())/(img_1.max()-img_1.min())*255)
print(img_2)
```

```python
img_equalizada = img_2

# Exiba a imagem original e a equalizada
plt.subplot(121), plt.imshow(image_gray, cmap='gray'), plt.title('Imagem Original')
plt.subplot(122), plt.imshow(img_equalizada, cmap='gray'), plt.title('Imagem Equalizada')
plt.show()
```
