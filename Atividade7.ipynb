```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
imagem = cv2.imread('/content/drive/MyDrive/imagens/img_JWST_04.png')
imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

#imagem = cv2.resize(imagem, (256, 256))

plt.imshow(imagem)
plt.axis('off')
plt.show()
```

```python
filterB = np.zeros_like(imagem)

filterB[:, :, 2] = imagem[:, :, 2]

plt.imshow(filterB)
plt.show()
```

```python
img_float = imagem.astype(np.float32)

pesos = [0, 0, 1]
img_mod = img_float.copy()

for i in range(3):
  img_mod[:, :, i] = img_float[:, :, i] * pesos[i]


img_res = np.clip(img_mod, 0, 255).astype(np.uint8)

limite = 70 # ajuste este valor conforme necessário
img_res[img_res[:, :, 2] < limite] = [0, 0, 0]

plt.imshow(img_res)
plt.show()
```
