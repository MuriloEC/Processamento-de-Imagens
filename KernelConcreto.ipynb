```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
imagem = cv2.imread('/content/drive/MyDrive/imagens/tomografia_concreto.jpg')
imagem2 = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

plt.imshow(imagem)
plt.show()


kernel = np.array([[-1,-1,-1],
                   [-1,8,-1],
                   [-1,-1,-1]])

img_filt = cv2.filter2D(imagem2, -1, kernel)

plt.imshow(img_filt)
plt.show()
```
