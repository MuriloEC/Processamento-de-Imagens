```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/drive/MyDrive/imagens/filter2d_dog.jpg', cv2.IMREAD_GRAYSCALE)

edges = cv2.Canny(image, 300, 350)

plt.subplot(121), plt.imshow(image, cmap='gray'), plt.title('Imagem Original')
plt.subplot(122), plt.imshow(edges, cmap='gray'), plt.title('Detecção de Borda')
plt.show()
```
